```python
import time
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service

web="https://www.worldometers.info/population/countries-in-africa-by-population/"
path=r"C:\Users\Administrator\OneDrive\Documents\Data_Entry\PythonDOCS\chromedriver.exe"

service=Service(executable_path=path)
driver=webdriver.Chrome(service=service)

driver.get(web)
time.sleep(2)
driver.maximize_window()
```

```python
countries=driver.find_elements(By.XPATH,"(//tbody/tr/td[2]/a)")
population=driver.find_elements(By.XPATH,"(//tbody/tr/td[3])")
yearly_change=driver.find_elements(By.XPATH,"(//tbody/tr/td[4])")
net_change=driver.find_elements(By.XPATH,"(//tbody/tr/td[5])")
density=driver.find_elements(By.XPATH,"(//tbody/tr/td[6])")
land_area=driver.find_elements(By.XPATH,"(//tbody/tr/td[7])")
migrants=driver.find_elements(By.XPATH,"(//tbody/tr/td[8])")
fertility_rate=driver.find_elements(By.XPATH,"(//tbody/tr/td[9])")
age=driver.find_elements(By.XPATH,"(//tbody/tr/td[10])")
urban_population=driver.find_elements(By.XPATH,"(//tbody/tr/td[11])")
world_share=driver.find_elements(By.XPATH,"(//tbody/tr/td[12])")
```

```python
population_results=[]
for i in range(len(countries)):
  temporary_data={'Country':countries[i].text,
                  'Population':population[i].text,
                  'Yearly change':yearly_change[i].text,
                  'Net change':net_change[i].text,
                  'Density':density[i].text,
                  'Land area':land_area[i].text,
                  'Migrants':migrants[i].text,
                  'Fertility rate':fertility_rate[i].text,
                  'Age':age[i].text,
                  'Urban population':urban_population[i].text,
                  'World share':world_share[i].text}
  population_results.append(temporary_data)
  
```

```python
df_data=pd.DataFrame(population_results)
print(df_data)
```
