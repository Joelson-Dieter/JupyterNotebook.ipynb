```python
import time
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service

web="https://www.flipkart.com/home-kitchen/kitchen-appliances/coffee-makers/pr?sid=j9e,m38,wqo&otracker=nmenu_sub_Appliances_0_Coffee%20Makers&otracker=nmenu_sub_TVs%20%26%20Appliances_0_Coffee%20Makers"
path=r"C:\Users\Administrator\OneDrive\Documents\Data_Entry\PythonDOCS\chromedriver.exe"

service=Service(executable_path=path)
driver=webdriver.Chrome(service=service)

driver.get(web)
time.sleep(2)

titles=[]
colors=[]
discounts=[]
prices=[]
```

```python
for i in range(155):

  title=driver.find_elements(By.XPATH,"(//a[@class='wjcEIp'])")
  color=driver.find_elements(By.XPATH,"(//div[@class='NqpwHC'])")
  discount=driver.find_elements(By.XPATH,"(//div[@class='UkUFwK'])")
  price=driver.find_elements(By.XPATH,"(//div[@class='Nx9bqj'])")

  for t in title:
    titles.append(t.text)
  for c in color:
    colors.append(c.text)
  for d in discount:
    discounts.append(d.text)
  for p in price:
    prices.append(p.text)

next_button=driver.find_element(By.XPATH,"(//a[@class='_9QVEpD'])")
next_button.click()
time.sleep(2)
```

```python
print(len(titles))
print(len(colors))
print(len(discounts))
print(len(prices))
```

```python
df=pd.DataFrame({'Titles':titles,
                 'Colors':colors,
                 'Discounts':discounts,
                 'Prices':prices})

print(df)
```
