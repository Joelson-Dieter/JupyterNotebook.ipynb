```python
import time
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
web="https://en.wikipedia.org/wiki/List_of_counties_of_Kenya_by_population"
path=r"C:\Users\Administrator\OneDrive\Documents\Data_Entry\PythonDOCS\chromedriver.exe"
service=Service(executable_path=path)
driver=webdriver.Chrome(service=service)
driver.get(web)
driver.maximize_window()
time.sleep(2)
```

```python
rank=driver.find_elements(By.XPATH,"(//tbody/tr/td[1])")
county=driver.find_elements(By.XPATH,"(//tbody/tr/td[2])")
population_2023=driver.find_elements(By.XPATH,"(//tbody/tr/td[3])")
population_2019=driver.find_elements(By.XPATH,"(//tbody/tr/td[4])")
population_2009=driver.find_elements(By.XPATH,"(//tbody/tr/td[5])")
population_1999=driver.find_elements(By.XPATH,"(//tbody/tr/td[6])")
population_1989=driver.find_elements(By.XPATH,"(//tbody/tr/td[7])")
population_1979=driver.find_elements(By.XPATH,"(//tbody/tr/td[8])")
percentage_total_2003=driver.find_elements(By.XPATH,"(//tbody/tr/td[6])")

```

```python
population_results=[]
for i in range(len(rank)):
  temporary_data={'Rank':rank[i].text,
                  'County':county[i].text,
                  'Population 2003':population_2023[i].text,
                  'Population 2019':population_2019[i].text,
                  'Population 2009':population_2009[i].text,
                  'Population 1999':population_1999[i].text,
                  'Population 1989':population_1989[i].text,
                  'Population_1979':population_1979[i].text,
                  'Percentage total':percentage_total_2003[i].text}
  population_results.append(temporary_data)
```

```python
df_data=pd.DataFrame(population_results)
print(df_data)
```
